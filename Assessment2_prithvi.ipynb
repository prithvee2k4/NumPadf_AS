Q11.

```python
student_performance_path = "/content/drive/MyDrive/DSAB13/CaseStudy/CS2/student_performance.csv"

student_performance_df = pd.read_csv(student_performance_path)

print("First 5 rows:")
display(student_performance_df.head())
print("\n")

print("Last 5 records:")
print(student_performance_df.tail())
print("\n")

print("number of rows and columns:")
print(student_performance_df.shape)
print("\n")

print("column names:")
print(student_performance_df.columns)
print("\n")

print("data types:")
print(student_performance_df.dtypes)
print("\n")

print("Statistical summary of the numerical columns:")
print(student_performance_df.describe())
```

Q12. Filtering and Sorting

```python
# 1. Find students whose Final_Score is greater than 75

result1 = student_performance_df[student_performance_df['Final_Score'] > 75]
print("Final_Score is greater than 75:")
print(result1)
print("\n")

# 2. Find students whose Attendance is below 75%

result2 = student_performance_df[student_performance_df['Attendance'] < 75]
print("Attendance is below 75%:")
print(result2)
print("\n")

# 3. Find Data Science students with Final_Score greater than 80

result3 = student_performance_df[
    (student_performance_df['Department'] == 'Data Science') &
    (student_performance_df['Final_Score'] > 80)
]

print("Final_Score greater than 80:")
print(result3)
print("\n")

#4. Sort all students by Final_Score from highest to lowest

print("Final_Score from highest to lowest:")
result4 = student_performance_df.sort_values(by='Final_Score', ascending=False)

print(result4)
```

Q13. Feature Creation

```python
student_performance_df['Average_Technical_Score'] = student_performance_df[
    ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)

print(student_performance_df)

conditions = [
    student_performance_df['Average_Technical_Score'] >= 90,
    student_performance_df['Average_Technical_Score'] >= 75,
    student_performance_df['Average_Technical_Score'] >= 60
]

choices = ['Excellent', 'Good', 'Average']

student_performance_df['Performance'] = np.select(
    conditions,
    choices,
    default='Needs Improvement'
)

print(student_performance_df)
```

Q14. GroupBy Analysis

```python
dept_summary = student_performance_df.groupby('Department').agg(
    Student_Count=('Student_ID', 'count'),
    Avg_Final_Score=('Final_Score', 'mean'),
    Max_Final_Score=('Final_Score', 'max'),
    Min_Final_Score=('Final_Score', 'min')
).sort_values('Avg_Final_Score', ascending=False)

print("Department Summary:")
print(dept_summary)
```

Q15. NumPy & Pandas Integration

```python
scores = student_performance_df['Final_Score'].dropna().values

print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Std:", scores.std())
print("25th Percentile:", np.percentile(scores, 25))
print("75th Percentile:", np.percentile(scores, 75))
print("\n")

print("Final_Score is above the overall mean:")
print(student_performance_df[student_performance_df['Final_Score'] > scores.mean()])
```

Q16. Student Performance Analysis

```python
print("Missing values:")
print(student_performance_df.isnull().sum())
print("\n")

print("Handle missing values:")
student_performance_df['Python_Score'] = student_performance_df['Python_Score'].fillna(0)
student_performance_df['SQL_Score'] = student_performance_df['SQL_Score'].fillna(0)
student_performance_df['Pandas_Score'] = student_performance_df['Pandas_Score'].fillna(0)
student_performance_df['NumPy_Score'] = student_performance_df['NumPy_Score'].fillna(0)
print(student_performance_df)
print("\n")
student_performance_df['Average_Score'] = student_performance_df[
    ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
].mean(axis=1)
print("\n")
student_performance_df['Pass_Status'] = np.where(student_performance_df['Average_Score'] >= 60, 'Pass', 'Fail')
print(student_performance_df)

#
```
